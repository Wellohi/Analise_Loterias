```python
import os
import pandas as pd
```

```python
base_dir = os.getcwd()
path_data = os.path.join(base_dir, '..', 'data', 'raw', 'Quina.csv')

df = pd.read_csv(path_data, sep=';')
df
```

# Feature engineering

- Criação de features para gerar insights e analises.


```python
# Transformar a string na coluna de dezenas em lista de inteiros para realocar em colunas separadas
dezenas_split = df['dezenas'].str.split(',', expand=True) # O expand=True transforma a lista resultante em novas colunas automaticamente

# Renomear as colunas 
dezenas_split.columns = [f'bola_{i+1}' for i in range(dezenas_split.shape[1])]

for col in dezenas_split.columns:
    dezenas_split[col] =  pd.to_numeric(dezenas_split[col])
    
df_final = pd.concat([df, dezenas_split], axis=1)

df_final = df_final.drop(columns=['dezenas'])

df_final
```

### Features estatísticas

- Soma
- Média
- Amplitude
- Quantidade de pares e ímpares

```python
colunas_bolas = [f'bola_{i+1}' for i in range(5)]

df_features = df_final.copy()

# Soma e Média: jogo foi mais 'alto' ou 'baixo'
df_features['soma'] = df_features[colunas_bolas].sum(axis=1)
df_features['media'] = df_features[colunas_bolas].mean(axis=1)
# Amplitude: Distância do maio para o menor número
df_features['amplitude'] = df_features[colunas_bolas].max(axis=1) - df_features[colunas_bolas].min(axis=1)
df_features['qtd_pares'] = df_features[colunas_bolas].apply(lambda x: sum(1 for n in x if n % 2 == 0), axis=1)
df_features['qtd_impares'] = 5 - df_features['qtd_pares']
 

df_features
```
